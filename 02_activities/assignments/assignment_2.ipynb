#### Starter Code for Question 1

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def is_duplicate(root: TreeNode) -> int:
    if not root:
        return -1
    
    seen = set()
    queue = [root]
    
    while queue:
        node = queue.pop(0)
        if node.val in seen:
            return node.val
        seen.add(node.val)
        
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
    
    return -1
```

```python
# Your answer here
We need to check if a string made up only of brackets ((), {}, []) is properly balanced. 
That means every opening bracket must have a matching closing bracket of the same type, 
and they must be closed in the correct order. 
If the sequence is valid, return True; otherwise, return False.
```


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


```python
# Your answer here
Input: s = "{[()()]}"
Output: True
```


-   Copy the solution your partner wrote. 


```python
# Your answer here
def is_valid_brackets(s: str) -> bool:
    stack = []

    pairs = {
        ')': '(',
        '}': '{',
        ']': '['
    }

    for i in s:
        if i in pairs.values():
            stack.append(i)

        elif i in pairs:
            if not stack or stack[-1] != pairs[i]:
                return False
            stack.pop()

    return len(stack) == 0
```


-   Explain why their solution works in your own words.


```python
# Your answer here
The stack keeps track of unmatched opening brackets. 
When a closing bracket appears, the code checks if it matches the most recent opening bracket (the stack top). 
If it does, that opening bracket is removed (popped). 
If not, the sequence is invalid. 
At the end, if the stack is empty, all brackets were matched correctly.
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Your answer here
- Time complexity: O(n), where n is the length of the string, since each character is processed once.
- Space complexity: O(n) in the worst case, if all characters are opening brackets and stored in the stack.
```
