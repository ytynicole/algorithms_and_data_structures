## Question One: First Duplicate in List
**Description**  
Given a list of integers, return the **first value that appears more than once**. If there are multiple duplicates, return the one that appears **first** in the list. If no duplicate exists, return `-1`.

**Examples**
```python
Input: nums = [3, 1, 4, 2, 5, 1, 6]
Output: 1
```
```python
Input: nums = [7, 8, 9, 10]
Output: -1
```
```python
Input: nums = [4, 5, 6, 4, 6]
Output: 4
```

**Question 1 Starter Code**

````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1
````

## Question Two: Valid Bracket Sequence
**Description**  
Given a string containing only the characters `'('`, `')'`, `'{'`, `'}'`, `'['`, and `']'`, determine if the input string is a **valid bracket sequence**.  
A string is valid if:
- Open brackets are closed by the same type of brackets, and
- Open brackets are closed in the correct order.

**Examples**
```python
Input: s = "([]{})"
Output: True
```
```python
Input: s = "([)]"
Output: False
```
```python
Input: s = "()[]{}"
Output: True
```
```python
Input: s = "[{]}"
Output: False
```

**Question 2 Starter Code**

````python
def is_valid_brackets(s: str) -> bool:
    matching = {')': '(', '}': '{', ']': '['}
    stack = []
    for ch in s:
        if ch in matching.values():  
            stack.append(ch)
        elif ch in matching:         
            if not stack or stack[-1] != matching[ch]:
                return False
            stack.pop()
        else:
            return False
    return not stack
````

## Question Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
```python
Input: nums = [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```
```python
Input: nums = [4, 0, 5, 0, 0, 6]
Output: [4, 5, 6, 0, 0, 0]
```


````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    """
    Move all zeros in nums to the end while preserving order of non-zero elements.
    Modifies nums in-place and returns it for convenience.
    """
    slow = 0
    for fast in range(len(nums)):
        if nums[fast] != 0:
            nums[slow] = nums[fast]
            slow += 1
    # fill remaining positions with zeros
    for i in range(slow, len(nums)):
        nums[i] = 0
    return nums
````

````python
# Your answer here
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1

# Example runs
if __name__ == "__main__":
    print(first_duplicate([3, 1, 4, 2, 5, 1, 6]))   
    print(first_duplicate([7, 8, 9, 10]))           
    print(first_duplicate([4, 5, 6, 4, 6]))        
    print(first_duplicate([2, 3, 5, 2, 3]))      
````


-   Explain why your solution works


````python
# Your answer here
As we iterate the list from left to right, we keep a set of values we've already encountered. 
The first time we encounter a value that is already in the set, that value’s second occurrence is the earliest second occurrence possible because we are scanning in order. 
Returning that value immediately satisfies the requirement to return the duplicate whose repeat appears first. 
If the loop completes without finding any repeated value, there are no duplicates and we return -1.
````


-   Explain the problem’s time and space complexity


````python
# Your answer here
Time complexity O(n) where n is the length of the list, because we perform a single pass and each set membership/add operation is amortized O(1).
Space complexity O(n) in the worst case to store the seen set when there are no duplicates.

````


-   Explain the thinking to an alternative solution (no coding required, but a classmate reading this should be able to code it up based off your text)


````python
# Your answer here
If the input guarantees that all numbers are positive and bounded by the list length (for example values in the range 1 to n), we can solve this in O(n) time and O(1) extra space by using the input array itself to mark visited indices: for each value v encountered, map it to index |v|-1 and negate the element at that index to mark it visited. 
````
